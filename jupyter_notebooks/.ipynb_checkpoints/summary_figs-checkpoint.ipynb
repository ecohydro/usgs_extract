```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Load the merged dataframe
df = pd.read_csv('merged_dataframe.csv')

print(f"Loaded merged dataframe with shape: {df.shape}")
print(f"Available columns: {list(df.columns)}")
print()

# Check if we have the expected columns
required_columns = ['number_rows', 'water_type_clean', 'year_start', 'year_end']
missing_columns = [col for col in required_columns if col not in df.columns]
if missing_columns:
    print(f"Missing required columns: {missing_columns}")
    print("Available columns that might work:")
    for col in df.columns:
        if any(keyword in col.lower() for keyword in ['year', 'water', 'number', 'count']):
            print(f"  - {col}")
    exit()

# Create decade column using vectorized operation
df['decade'] = (df['year_start'] // 10) * 10

# Remove rows with no decade assignment
df_clean = df.dropna(subset=['decade', 'water_type_clean'])
print(f"After removing rows with missing decade/water_type: {df_clean.shape}")

# Check decade range
print(f"Decade range: {df_clean['decade'].min()} to {df_clean['decade'].max()}")
print(f"Unique decades: {sorted(df_clean['decade'].unique())}")
print()

# Group by decade and water type, sum the measurements
decade_water_summary = df_clean.groupby(['decade', 'water_type_clean']).agg({
    'number_rows': 'sum'
}).reset_index()

print("Sample of decade-water type summary:")
print(decade_water_summary.head(10))
print()

# Pivot for easier plotting
pivot_data = decade_water_summary.pivot(index='decade', columns='water_type_clean', values='number_rows')
pivot_data = pivot_data.fillna(0)

# Remove unwanted water type categories from the data
categories_to_exclude = ['Other', 'Not water related']
pivot_data = pivot_data.drop(columns=[col for col in categories_to_exclude if col in pivot_data.columns])

# Remove decades with zero values and exclude 1980
pivot_data = pivot_data[pivot_data.sum(axis=1) > 0]
pivot_data = pivot_data[pivot_data.index != 1980]

print("Water types found:")
print(list(pivot_data.columns))
print()

# Define colors to match your legend
color_palette = {
    'Stream discharge': '#4472C4',   # Blue
    'Groundwater': '#E06C75',        # Pink/Red
    'Reservoir': '#70AD47',          # Green
    'Irrigation': '#FFC000',         # Yellow/Gold
    'Springs': '#5B9BD5',            # Light Blue
    'Precipitation': '#C5504B',      # Dark Red/Maroon
    'Water quality': '#264478',      # Dark Blue
    'Not water related': '#808080',  # Gray (fallback)
    'Other': '#D62728'               # Dark red (fallback)
}

# Get colors for available water types
# Define the legend order (Stream discharge at top) - excluding unwanted categories
legend_order = [
    'Stream discharge',
    'Groundwater', 
    'Reservoir',
    'Irrigation',
    'Springs',
    'Precipitation',
    'Water quality'
]

# Keep only water types that exist in the data, in the specified order
available_types = [wt for wt in legend_order if wt in pivot_data.columns]

# Add any water types not in our predefined order at the end
remaining_types = [wt for wt in pivot_data.columns if wt not in legend_order]
available_types.extend(remaining_types)
colors = [color_palette.get(water_type, '#CCCCCC') for water_type in available_types]

# Create the stacked bar chart with wider bars
plt.figure(figsize=(14, 8))

# Create stacked bars with increased width
bar_width = 6  # Make bars much wider
bottom = np.zeros(len(pivot_data))
bars = []

for i, water_type in enumerate(available_types):
    if water_type in pivot_data.columns:  # Check if this water type exists in data
        bar = plt.bar(pivot_data.index, pivot_data[water_type], 
                      width=bar_width,  # Add width parameter
                      bottom=bottom, label=water_type, 
                      color=colors[i], alpha=0.8)
        bars.append(bar)
        bottom += pivot_data[water_type]

# Customize the plot
plt.ylabel('Estimated number of measurements', fontsize=12)

# Set x-axis to show every decade
plt.xticks(pivot_data.index, [str(int(decade)) for decade in pivot_data.index], rotation=0)

# Format y-axis with thousands separator
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x:,.0f}'))

# Add legend outside the plot area
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

# Add grid for better readability
plt.grid(axis='y', alpha=0.3, linestyle='-', linewidth=0.5)

# Add annotations for early decades with small values
early_decades_to_annotate = [1810, 1840, 1850, 1860]
annotation_heights = [25000, 40000, 90000, 120000]  # More staggered heights

for i, decade in enumerate(early_decades_to_annotate):
    if decade in pivot_data.index:
        decade_total = pivot_data.loc[decade].sum()
        if decade_total > 0:
            # Get the dominant water type for this decade
            decade_data = pivot_data.loc[decade]
            max_water_type = decade_data.idxmax()
            
            # Add annotation with arrow, staggered heights
            plt.annotate(f'{int(decade_total):,}\n({max_water_type})', 
                        xy=(decade, decade_total), 
                        xytext=(decade, annotation_heights[i]),
                        ha='center', va='bottom',
                        fontsize=9,
                        arrowprops=dict(arrowstyle='->', color='black', lw=0.8))

# Adjust layout to prevent legend cutoff
plt.tight_layout()

# Show summary statistics
print("=== SUMMARY STATISTICS ===")
print(f"Total records plotted: {pivot_data.sum().sum():,.0f}")
print(f"Decades covered: {len(pivot_data)} ({int(pivot_data.index.min())}-{int(pivot_data.index.max())})")
print()

print("Records by water type:")
water_type_totals = pivot_data.sum().sort_values(ascending=False)
for water_type, total in water_type_totals.items():
    print(f"  {water_type}: {total:,.0f}")
print()

print("Records by decade:")
decade_totals = pivot_data.sum(axis=1).sort_index()
for decade, total in decade_totals.items():
    print(f"  {int(decade)}s: {total:,.0f}")

# Save the figure
plt.savefig('water_type_by_decade.png', dpi=300, bbox_inches='tight')
plt.show()

# Optionally save the summary data
pivot_data.to_csv('decade_water_type_summary.csv')
print(f"\nChart saved as 'water_type_by_decade.png'")
print(f"Summary data saved as 'decade_water_type_summary.csv'")
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
from pathlib import Path

# Set your image directory path
image_dir = Path(r"C:\Users\aeliz\Dropbox\Documents\Bren School PhD\Dissertation\Chapter 2 USGS Digitization\figures\example tables\final")

# Get all image files from the directory (common image formats)
image_extensions = ['.png', '.jpg', '.jpeg', '.tiff', '.tif', '.bmp']
image_files = []
for ext in image_extensions:
    image_files.extend(list(image_dir.glob(f"*{ext}")) + list(image_dir.glob(f"*{ext.upper()}")))

# Remove duplicates and sort the files to ensure consistent ordering
image_files = sorted(list(set(image_files)))

# Check if we have at least 9 images
if len(image_files) < 9:
    print(f"Warning: Only {len(image_files)} images found. Need 9 images for 3x3 grid.")
    print("Available images:")
    for i, img in enumerate(image_files):
        print(f"{i+1}: {img.name}")
else:
    print(f"Found {len(image_files)} images. Using first 9 for the panel.")

# Create the 3x3 subplot figure with minimal spacing
fig, axes = plt.subplots(3, 3, figsize=(15, 15))

# Remove ALL space between subplots
plt.subplots_adjust(hspace=0, wspace=0, left=0, right=1, top=1, bottom=0)

# Display images in the grid
for i in range(3):
    for j in range(3):
        idx = i * 3 + j
        ax = axes[i, j]
        
        if idx < len(image_files):
            # Load and display image
            img = mpimg.imread(image_files[idx])
            
            # Use 'fill' aspect to make all images fill their subplot completely
            # This will stretch images to fit the subplot dimensions
            #ax.imshow(img, aspect='equal')
            
            # Alternative: Use 'auto' aspect for less distortion but potentially different sizes
            ax.imshow(img, aspect='auto')
            
        else:
            # Hide empty subplots if less than 9 images
            ax.axis('off')
            ax.text(0.5, 0.5, 'No Image', ha='center', va='center', transform=ax.transAxes)
        
        # Remove axis ticks, labels, and spines for completely clean look
        ax.set_xticks([])
        ax.set_yticks([])
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.spines['bottom'].set_visible(False)
        ax.spines['left'].set_visible(False)

# Save the figure with tight bounding box
plt.savefig('image_panel_3x3.png', dpi=300, bbox_inches='tight', pad_inches=0)

# Display the figure
plt.show()

# Optional: Print list of images being used
print("\nImages displayed in the panel:")
for i, img_file in enumerate(image_files[:9]):
    row = i // 3 + 1
    col = i % 3 + 1
    print(f"Position ({row},{col}): {img_file.name}")
```

## Figure 1 — Example plot

> Replace the code below with your actual plotting logic for Figure 1.
> The snippet shows a generic pattern with matplotlib using a colorblind-friendly palette.


````
import matplotlib.pyplot as plt

# Choose the table that drives Figure 1
# Replace 'df_summary' with the correct table name from `tables`
df = tables.get("df_summary")
if df is None:
    raise KeyError("Expected a table named 'df_summary' in artifacts. Update the key or export step.")

# Example placeholder plot (edit to your needs)
plt.figure(figsize=(6,4), dpi=150)
# Example: bar plot of counts by site (customize with your real columns)
if {'site_id','n'}.issubset(df.columns):
    df_group = df.groupby('site_id')['n'].sum().sort_values()
    df_group.plot(kind='bar')
    plt.ylabel("Count")
    plt.title("Example: Counts by Site")
    plt.tight_layout()
    plt.show()
else:
    print("Update this cell to match your actual df columns.")
````

## Figure 2 — Example plot

> Replace the code below with your actual plotting logic for Figure 2.


````
import matplotlib.pyplot as plt

# Choose the table that drives Figure 2
# Replace 'df_events' with the correct table name from `tables`
df = tables.get("df_events")
if df is None:
    raise KeyError("Expected a table named 'df_events' in artifacts. Update the key or export step.")

# Example placeholder plot (edit to your needs)
plt.figure(figsize=(6,4), dpi=150)
# Example: histogram of an event metric (e.g., 'Tr_hours'); adjust to your real column
candidate_cols = [c for c in df.columns if 'Tr' in c or 'duration' in c or 'mm' in c]
if candidate_cols:
    col = candidate_cols[0]
    df[col].plot(kind='hist', bins=30)
    plt.xlabel(col)
    plt.title(f"Example: Distribution of {col}")
    plt.tight_layout()
    plt.show()
else:
    print("Update this cell to pick the right metric column.")
````

```python
import json
import lxml
import pandas as pd
from pathlib import Path

# Your JSON file path
path = Path(r"C:\Users\aeliz\Dropbox\Chap2_USGS\2252_728 example\2252_page_728.json")

# Load JSON
with path.open("r", encoding="utf-8") as f:
    j = json.load(f)

# Extract tables
tables = []
for chunk in j.get("result", {}).get("chunks", []):
    for block in chunk.get("blocks", []):
        if block.get("type") == "Table" and block.get("content"):
            dfs = pd.read_html(block["content"])
            tables.extend(dfs)

print(f"Loaded {len(tables)} table(s).")
if tables:
    print(tables[0].head())

```

```python
import pandas as pd
import requests
import json
```

```python
key = "sk-a4152196f16845598b062a9a1fc82f38"
def chat_with_model(token, promp, model):
    url = 'https://llm.grit.ucsb.edu/api/chat/completions'
    headers = {
        'Authorization': f'Bearer {token}',
        'Content-Type': 'application/json',
        'response_format': 'json_object'
    }
    
    data = {
      "model": model,
      "messages": [
        {
          "role": "user",
          "content": promp
        }
      ]
    }
    response = requests.post(url, headers=headers, json=data)
    return response.json()
```

```python

filep = "/Users/yiboliang/Downloads/JSON Context Extraction Tables Only/json/25076_page_32.json"

with open(filep) as json_file:
    data = json.load(json_file)['result']['chunks']
```

```python
parsed = ""
for bbox in data:
    for block in bbox['blocks']:
        if block['type'] == 'Header':
            parsed = parsed + block['content']
        elif block['type'] == 'Table':
            parsed = parsed + block['content']
```

```python
model = "deepseek-r1:latest"
prompt = f"""
            You are an assistant that will summarize html tables. Be as detailed as possible in your summaries.
            Here is the data:\n{parsed}\n
        
            """
```

```python
res = chat_with_model(key, prompt, model)
```

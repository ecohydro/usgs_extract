# Table Recovery

```python
import os
import cv2
from paddleocr import PPStructure,save_structure_res
from paddleocr.ppstructure.recovery.recovery_to_doc import sorted_layout_boxes, convert_info_docx

# English image
table_engine = PPStructure(image_orientation=True, recovery=True, lang='en')
save_folder = "output"
tables_path = 'tables'
pages = os.listdir(tables_path)
pages_paths = []
for page in pages:
    if page[-4:] == ".png":
        pages_paths.append(os.path.join(tables_path, page))
for path in pages_paths:
    images = os.listdir(path)
    for image in images:
        try:
            name = os.path.basename(path).split('.')[0] + "_" + image[:-4]
            img_path = os.path.join(path, image)
            img = cv2.imread(img_path)
            result = table_engine(img)
            save_structure_res(result, save_folder, name)
            h, w, _ = img.shape
            res = sorted_layout_boxes(result, w)
            convert_info_docx(img, res, save_folder, name)
        except Exception as e:
            print(f"Found Exception {e}")
```
